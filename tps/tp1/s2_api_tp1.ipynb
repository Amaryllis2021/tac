# APIs: requêtes HTTP

## Imports

```python
import json
import requests
import os
```

## Utiliser Open Library (livres par sujet): https://openlibrary.org/developers/api 

```python
subject = "Bruxelles"

"""Retrieve books from Open Library"""
url = f"https://openlibrary.org/subjects/{subject}.json"
resp = requests.get(url)

try:
    data = json.loads(resp.text)
except json.JSONDecodeError:
    print(resp.text)

for work in data['works'][:5]:
    title = work['title']
    authors = ", ".join(a['name'] for a in work.get('authors', []))
    print(f"{title} ({authors})")
```
